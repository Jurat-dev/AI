```python
import pandas as pd
```

ONE-HOT ENCODING

```python
dummies = pd.get_dummies(df['fuel-system'], prefix='system',dtype=int)
```

```python
dummies
```

```python
dummies.head()
```

```python
df=pd.concat([df.drop('fuel-system', axis=1), dummies], axis=1)
```

```python
df.info()
```

```python
df.head()
```

```python
df["normalized-losses"] = df["normalized-losses"].fillna(df["normalized-losses"].mean())
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df["num-of-doors"] = df["num-of-doors"].fillna(df["num-of-doors"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.dropna(inplace=True)
```
