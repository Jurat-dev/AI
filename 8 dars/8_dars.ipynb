```python
import pandas as pd
```

```python
df=pd.read_csv('buyurtmalar_mashq.csv')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
for col in df.columns:
    if df[col].isnull().any():
          if pd.api.types.is_numeric_dtype(df[col]):
            df[col] = df[col].fillna(df[col].mean())      # son bo'lsa: mean
          else:
            df[col] = df[col].fillna(df[col].mode()[0])   # matn bo'lsa: mode
```
