```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
```

```python
df = pd.read_csv("buyurtmalar_mashq.csv")
print(df.shape)
```

```python
for col in df.columns:
    if df[col].isnull().any():
        print(col, "-", df[col].isnull().sum(), "ta bosh joy")
```

```python
#1-usul: Mean
a = df.copy()
for col in a.columns:
    if a[col].isnull().any() and pd.api.types.is_numeric_dtype(a[col]):
        a[col] = a[col].fillna(a[col].mean())
print("\n1) Mean:", a.select_dtypes("number").isnull().sum().sum(), "ta son bosh joy qoldi")

#2-usul: Median
a = df.copy()
for col in a.columns:
    if a[col].isnull().any() and pd.api.types.is_numeric_dtype(a[col]):
        a[col] = a[col].fillna(a[col].median())
print("2) Median:", a.select_dtypes("number").isnull().sum().sum(), "ta son bosh joy qoldi")

#3-usul: Mode 
a = df.copy()
for col in a.columns:
    if a[col].isnull().any():
        a[col] = a[col].fillna(a[col].mode()[0])
print("3) Mode:", a.isnull().sum().sum(), "ta bosh joy qoldi")
```

```python
#4-usul: Fixed
a = df.copy()
for col in a.columns:
    if a[col].isnull().any():
        if pd.api.types.is_numeric_dtype(a[col]):
            a[col] = a[col].fillna(0)
        else:
            a[col] = a[col].fillna("Nomalum")
print("4) Fixed:", a.isnull().sum().sum(), "ta bosh joy qoldi")

#5-usul: Drop
a = df.copy()
for col in a.columns:
    if a[col].isnull().any():
        a = a.dropna(subset=[col])
print("5) Drop:", len(df) - len(a), "ta qator ochirildi, qoldi:", len(a))
```

# Scaling



```python
toza = df.copy()
son_ustunlar = []
for col in toza.columns:
    if pd.api.types.is_numeric_dtype(toza[col]):
        son_ustunlar.append(col)    #append = bosh qator qoshayabdi
print("\nSon ustunlar:", son_ustunlar)

#1-usul: MinMaxScaler (0 dan 1 gacha)
minmax = toza.copy()
scaler = MinMaxScaler()
for col in son_ustunlar:
    minmax[col] = scaler.fit_transform(minmax[[col]])
print("\nMinMaxScaler:")
print(minmax[son_ustunlar].head())

#2-usul: StandardScaler (ortacha 0, std 1)
standard = toza.copy()
scaler = StandardScaler()
for col in son_ustunlar:
    standard[col] = scaler.fit_transform(standard[[col]])
print("\nStandardScaler:")
print(standard[son_ustunlar].head())

#3-usul: RobustScaler (median va IQR bilan, outlierlarga chidamli)
robust = toza.copy()
scaler = RobustScaler()
for col in son_ustunlar:
    robust[col] = scaler.fit_transform(robust[[col]])
print("\nRobustScaler:")
print(robust[son_ustunlar].head())
```
